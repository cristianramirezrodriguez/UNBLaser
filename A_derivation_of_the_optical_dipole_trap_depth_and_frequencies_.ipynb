# A derivation of the optical dipole trap depth and frequencies by Cristian Ramirez Rodriguez

The dipole potential produced by a single gaussian beam is:
$$U_{dip} = -\frac{\alpha}{2}|E(\vec{r})|^2$$

From Milonni and Eberly, the Electric Field of a Gaussian Beam (Gaussian Beam Solution of Paraxial Wave Equation) is
$$E(\vec{r})=A\frac{W_0}{W(z)}\exp({i[kz-\arctan(z/z_R)]})\exp({ik((x^2+y^2)/2R(z))})\exp({-i\omega t})\exp({-(x^2+y^2)/(W(z))^2})$$

The dipole potential is then
$$U_{dip}=|A|^2\frac{\alpha W_0^2}{2(W(z))^2}\exp({-2(x^2+y^2)/(W(z))^2})$$

where
$$W(z)= W_0\sqrt{1+\frac{z}{z_R}}$$

and $$|A|^2 = \frac{2I_0}{c\epsilon_0}$$

We define $I_0$ in terms of the power ramp through

$$P=\int I_0 e^{-2r^2/W_0^2}dr = I_0\frac{\pi}{2}W_0$$

This allows us to express our trapping potential as

$$U_{dip}(x,y,z)=-U_0\frac{1}{(1+\frac{z^2}{z_R^2})}\exp(\frac{-2(x^2+y^2)}{W_0^2(1+\frac{z^2}{z_R^2})})$$

with (all quantities in SI units)

$$U_0 = U(0,0,0) = \frac{2\alpha_{SI} P}{\epsilon_0 c \pi W_0^2}$$

This is equivalent to Gupta et. al's power for a single unmodulated beam (in Gupta et. al, the atomic polarization is not in SI units, and unit analysis confirms $\alpha_{paper} = \frac{\alpha_{SI}}{c\epsilon_0}$).

To find the trap frequencies we must equate $U_{dip}(x,y,z)$ to the simple harmonic potential, $U_{harm} =\frac{1}{2}m\omega^2 (x^2 + y^2 +z^2)$

$$-U_0\frac{1}{(1+\frac{z^2}{z_R^2})}\exp(\frac{-2(x^2+y^2)}{W_0^2(1+\frac{z^2}{z_R^2})})=\frac{1}{2}m(\omega_x^2x^2 + \omega_y^2y^2 + \omega_z^2z^2)-U_0$$

Applying the Taylor Expansion $e^x \approx 1 + x$:

$-\frac{U_0}{(1+\frac{z^2}{z_R^2})}-\frac{U_0}{(1+\frac{z^2}{z_R^2})}(\frac{2(x^2+y^2)}{W_0^2(1+\frac{z^2}{z_R^2})})=\frac{1}{2}m(\omega_x^2x^2 + \omega_y^2y^2 + \omega_z^2z^2)-U_0$

Using another Taylor expansion for small z, $\frac{1}{1+\frac{z^2}{z_R^2}} ≈ 1 - \frac{z^2}{z_R^2}$

$-U_0 + \frac{U_0z^2}{z_R^2}-\frac{U_0 2(x^2+y^2)}{W_0^2(1+\frac{z^2}{z_R^2})}+\frac{U_0 z^2 2(x^2+y^2)}{z_R^2 W_0^2(1+\frac{z^2}{z_R^2})}=-U_0 + \frac{U_0z^2}{z_R^2}-\frac{U_0 2(x^2+y^2)}{W_0^2}+\frac{U_0 z^2 2(x^2+y^2)}{z_R^2 W_0^2}+\frac{U_0 z^2 2(x^2+y^2)}{z_R^2 W_0^2}-\frac{U_0 z^4 2(x^2+y^2)}{z_R^4 W_0^2}=\frac{1}{2}m(\omega_x^2x^2 + \omega_y^2y^2 + \omega_z^2z^2)-U_0$

For the radial directions:

$$U(x,y,0) = -U_0 - \frac{U_0 2(x^2+y^2)}{W_0^2}=\frac{1}{2}m(\omega_x^2x^2 + \omega_y^2y^2)-U_0$$

$$\omega_x^2 = \frac{4U_0}{mW_0^2}, \omega_y^2 = \frac{4U_0}{mW_0^2}$$ as in Gupta et. al when $f_\omega(h/W_0)=1$

For the z direction:

$$-U_0 + \frac{U_0z^2}{z_R^2}=\frac{m}{2}\omega_z^2z^2-U_0$$

$$\omega_z^2 = \frac{2U_0}{mW_0^2z_R^2}$$


The equations of state are

$$\dot{N}=-(8\pi na^2\sqrt{\frac{8k_bT}{\pi m}}(\frac{U}{kT}-4)\exp{(-\frac{U}{kT})}+\frac{K_3 n^2}{3 \sqrt{3}}+\Gamma_{bg})N$$

$$\dot{T} = -(\frac{8\pi na^2\sqrt{\frac{8k_bT}{\pi m}}(\frac{U}{kT}-4)\exp{(-\frac{U}{kT})}}{3}(\frac{U}{kT} + \frac{\frac{U}{kT}-5}{\frac{U}{kT}-4} -3)-\frac{\frac{K_3 n^2}{3 \sqrt{3}}}{3}-\frac{\dot{\bar{\omega}}}{\bar{\omega}})T+\frac{\Gamma_{sc}E_r}{3k_B}$$

$n \approx \frac{N}{\pi W_0^2}$
